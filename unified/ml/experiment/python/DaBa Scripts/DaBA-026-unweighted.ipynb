```python
import pandas as pd
import numpy as np
import math
from scipy.stats import variation  

# This is Statistics function that input dataframe and target value, then
# removes outliers from dataframe and 
# output a dataframe including all required statistics.

def statistics(data, target):
    

    def outliers_IQR(data, t=3):      # This function removes outliers from given datafrme
        threshold = t
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1
        dataf = data[~((data <(Q1 - threshold * IQR)) | (data > (Q3 + threshold * IQR))).any(axis=1)]
        return dataf

    def stat_table(data, target):     # This function caculates statistics
        
        sdt = data.groupby("RIDAGEYR")[target].agg(['mean','std',('5th',lambda x: np.percentile(x, 5)),
                                                    ('25th',lambda x: np.percentile(x, 25)),
                                                    ('50th',lambda x: np.percentile(x, 50)),
                                                    ('75th',lambda x: np.percentile(x, 75)),
                                                    ('95th',lambda x: np.percentile(x, 95)),
                                                    'var'])
        return sdt

    odt = outliers_IQR(data, 3)
    sdt = stat_table(odt, target)
    return sdt
```
