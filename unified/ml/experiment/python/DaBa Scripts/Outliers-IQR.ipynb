```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import ExcelWriter
from pandas import ExcelFile
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
```

```python
address='NHANES 21SEPT2015.xlsx'
df=pd.read_excel(address, sheet_name='DXA_Imp_1')
```

```python
feature_set_1=['RIDAGEYR','BMXCALF','bmxbmi']
df_calf = df[feature_set_1]
df_calf = df_calf.dropna(axis=0)
df_calf.shape
```

```python
# This function removes outliers with IQR approach from the input data frame 
# and returns the outliers and the updated data frame
import numpy as np
from scipy import stats

def outliers_IQR(data, t=1.5):

    threshold = t
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    outliers = data[((data <(Q1 - threshold * IQR)) | (data > (Q3 + threshold * IQR))).any(axis=1)]
    dataf = data[~((data <(Q1 - threshold * IQR)) | (data > (Q3 + threshold * IQR))).any(axis=1)]
    return dataf, outliers
```

```python
dataf, outliers = outliers_IQR(df_calf,3)
```

```python
print("Data frame after removing outliers:\n", dataf)
```
