```python
import pandas as pd
import numpy as np
import math
import statsmodels.stats.weightstats as smw
```

```python
#This function performs a pairwise t-test considering unequal standard deviation with or without weighting. 
#Inputs= data1, data 2: two datasets which contains the target columns that should be compared
#target: target column name , 
#weight: If weighting system existed, name of weighting columns should be imported, otherwise test with no weighting is performed
#Outputs: pvalue: results of t-test. pvalue, tscore will be printed as well.  

def ttest_w(data1,data2, target, weight=0):
    
    var1=data1[target]
    var2=data2[target]
    
    if weight!=0:

        w1=data1[weight] 
        w2=data2[target]
        d1 = smw.DescrStatsW(var1, weights=w1)
        d2 = smw.DescrStatsW(var2, weights=w2)
        pvalue=(smw.ttest_ind(var1, var2, weights=(d1.weights,d2.weights), usevar='unequal'))[1]
        tscore=(smw.ttest_ind(var1, var2, weights=(d1.weights,d2.weights), usevar='unequal'))[0]

    else:
        pvalue=(smw.ttest_ind(var1, var2, usevar='unequal'))[1]
        tscore=(smw.ttest_ind(var1, var2, usevar='unequal'))[0]
        
    print('pvalue = %f' % pvalue)
    print('tscore = %f' % tscore)

    if pvalue<0.05:
        print('Significant difference in mean values are observed')
    
    return pvalue
```
