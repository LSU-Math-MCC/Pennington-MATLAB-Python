```python
address='NHANES 21SEPT2015.xlsx'
df=pd.read_excel(address, sheet_name='DXA_Imp_1')
```

```python
feature_set_1=['RIDAGEYR','BMXCALF','bmxbmi']
df_calf=df[feature_set_1]
df_calf
```

```python
df_calf = df_calf.dropna(axis=0)
df_calf.isnull().sum()
```

```python
# This function removes outliers with z-score approach from the input data frame 
# and returns the outliers and the updated data frame
import numpy as np
from scipy import stats

def outliers_z_score(data, t):

    threshold = t
    z_scores = np.abs(stats.zscore(data))
    
    outliers = np.where(z_scores > threshold)
    data = data[(z_scores < threshold).all(axis=1)]
    
    return outliers, data
```

```python
outliers, data = outliers_z_score(df_calf,3)
```
