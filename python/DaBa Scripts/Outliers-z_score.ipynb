```python
# This function removes outliers with z-score approach from the input data frame 
# and returns the outliers and the updated data frame
import numpy as np
from scipy import stats

def outliers_z_score(data, t=3):

    threshold = t
    try:
        z_scores = np.abs(stats.zscore(data))
    except ZeroDivisionError:
        print('devided by zero')    
    outlier_index = np.where(z_scores > threshold)
    
    outliers = data[~(z_scores < threshold).all(axis=1)]
    
    dataf = data[(z_scores < threshold).all(axis=1)]
    
    return outliers, dataf
  
```
