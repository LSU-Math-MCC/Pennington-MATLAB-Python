```python
import pandas as pd
import numpy as np
import math
import statsmodels.stats.weightstats as sm

def statistics(data, target, weight):
    
    def outliers_IQR(data, t=3):
        threshold = t
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1
        dataf = data[~((data <(Q1 - threshold * IQR)) | (data > (Q3 + threshold * IQR))).any(axis=1)]
        return dataf

    def stat_table(data, target,weight):

        Age = pd.unique(data["RIDAGEYR"])

        dff = pd.DataFrame({'Age':sorted(Age)})
        dff = dff.set_index('Age')
        
        groupedAges = data.groupby("RIDAGEYR")
        
        dff['Mean'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).mean)
        dff['Std'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).std)
        dff['Var'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).var)

        dff['25th'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).quantile(0.25))
        dff['50th'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).quantile(0.50))
        dff['75th'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).quantile(0.75))
        dff['95th'] = groupedAges.apply(lambda x: sm.DescrStatsW(x[target], x[weight]).quantile(0.95))

        return dff

    pdt = outliers_IQR(data, 3)
    fdt = stat_table(pdt, target, weight)
    print(fdt)
```

```python
address='Input-df.xlsx'
df=pd.read_excel(address)
```

```python
statistics(df,'BMXCALF','WTMEC8YR' )
```
